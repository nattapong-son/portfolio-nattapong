```python
class Introduction:
    def __init__(self, first_name, last_name, nick_name):
        self.first_name = first_name
        self.last_name = last_name
        self.nick_name = nick_name

    def call_my_fullname(self):
        return f"{self.first_name} {self.last_name}"

    def call_my_nickname(self):
        return self.nick_name


introduce = Introduction("Nattapong", "Songsaksri", "Weiss")
print(introduce.call_my_fullname())
print(introduce.call_my_nickname())
```
